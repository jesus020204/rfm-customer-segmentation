```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df_rfm = pd.read_csv('rfm_resultados.csv')

variables_rfm = df_rfm[['recencia_dias', 'frecuencia', 'valor_monetario']]
matriz_corr = variables_rfm.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(matriz_corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Matriz de Correlación: Recencia, Frecuencia y Monetización')
plt.show()

plt.figure(figsize=(10, 7))
sns.scatterplot(
    data=df_rfm, 
    x='frecuencia', 
    y='valor_monetario', 
    hue='segmento_cliente', 
    palette='viridis', 
    alpha=0.7
)
plt.title('Distribución de Clientes: Frecuencia vs Valor Monetario')
plt.xlabel('Frecuencia (Total de Compras)')
plt.ylabel('Valor Monetario ($)')
plt.legend(title='Segmento RFM')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
